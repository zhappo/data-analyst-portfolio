## Análise de Dados Faltantes (Missing Data)

```python
# Percentual de nulos por coluna
missing = (df_raw.isnull().sum() / len(df_raw) * 100).sort_values(ascending=False)
print("Dados Faltantes (%):")
print(missing[missing > 0])

# Visualiza
fig = px.bar(
    x=missing[missing > 0].values,
    y=missing[missing > 0].index,
    orientation='h',
    title='Percentual de Dados Faltantes',
    labels={'x': '% Nulo', 'y': 'Coluna'}
)
fig.show()
```

## Limpeza de Dados

### Problemas Identificados:
- Price: min=1€, max=1.38B€ (outliers extremos)
- TotalArea: valores negativos e absurdos (992M m²)
- NumberOfBedrooms: máximo 21 (possivéis erros de entrada)
- Mixed types em 8 colunas (conversão necessária)

### Estratégia:
1. Remove preços < 10k€ e > 5M€
2. Remove áreas negativas ou > 100k m²
3. Remove dormitórios > 10
4. Cria coluna Price/m²

```python
# Cria cópia independente para limpeza
df_clean = df_raw.copy()

print(f"Linhas antes: {len(df_clean)}")

# 1. Preços: 10k a 5M
df_clean = df_clean[(df_clean['Price'] >= 10000) & (df_clean['Price'] <= 5000000)]

# 2. Área total: > 0 e < 100k m²
df_clean = df_clean[(df_clean['TotalArea'] > 0) & (df_clean['TotalArea'] <= 100000)]

# 3. Dormitórios: 0-10
df_clean = df_clean[(df_clean['NumberOfBedrooms'] >= 0) & (df_clean['NumberOfBedrooms'] <= 10)]

# 4. Remove linhas com Price ou TotalArea nulos (essenciais)
df_clean = df_clean.dropna(subset=['Price', 'TotalArea'])

print(f"Linhas após limpeza: {len(df_clean)}")
print(f"Linhas removidas: {len(df_raw) - len(df_clean)} ({(len(df_raw) - len(df_clean))/len(df_raw)*100:.1f}%)")
```

```python
# Cria coluna Price/m²
df_clean['PricePerM2'] = df_clean['Price'] / df_clean['TotalArea']

print(df_clean[['Price', 'TotalArea', 'PricePerM2']].describe())
```

## 📊 Análise Exploratória (EDA)

### Distribuição de Preços

```python
fig = px.histogram(
    df_clean,
    x='Price',
    nbins=50,
    title='Distribuição de Preços de Imóveis',
    labels={'Price': 'Preço (€)'},
    color_discrete_sequence=['#1f77b4']
)
fig.update_layout(showlegend=False)
fig.show()

print(f"Preço Mediano: €{df_clean['Price'].median():,.0f}")
print(f"Preço Médio: €{df_clean['Price'].mean():,.0f}")
```

### Preços por Tipo de Imóvel

```python
fig = px.box(
    df_clean,
    x='Type',
    y='Price',
    title='Distribuição de Preços por Tipo de Imóvel',
    labels={'Price': 'Preço (€)', 'Type': 'Tipo'},
    color='Type'
)
fig.show()

print("\nPreço Médio por Tipo:")
print(df_clean.groupby('Type')['Price'].agg(['mean', 'median', 'count']).sort_values('median', ascending=False))

# Calcula premium de Casa vs Apartamento
if 'House' in df_clean['Type'].values and 'Apartment' in df_clean['Type'].values:
    house_price = df_clean[df_clean['Type'] == 'House']['Price'].median()
    apartment_price = df_clean[df_clean['Type'] == 'Apartment']['Price'].median()
    premium = ((house_price - apartment_price) / apartment_price) * 100
    
    print(f"\nCasa vs Apartamento:")
    print(f"Casa mediana: €{house_price:,.0f}")
    print(f"Apartamento mediano: €{apartment_price:,.0f}")
    print(f"Premium Casa: {premium:.1f}% {'mais caro' if premium > 0 else 'mais barato'}")
```

### Preços por Distrito (Top 10)

```python
district_stats = df_clean.groupby('District').agg({
    'Price': ['median', 'mean', 'count']
}).round(0)
district_stats.columns = ['Mediana', 'Média', 'Qtd_Imóveis']
district_stats = district_stats.sort_values('Mediana', ascending=False).head(10)

fig = px.bar(
    district_stats.reset_index(),
    x='District',
    y='Mediana',
    title='Top 10 Distritos - Preço Mediano',
    labels={'Mediana': 'Preço Mediano (€)', 'District': 'Distrito'},
    color='Mediana',
    color_continuous_scale='Viridis'
)
fig.update_xaxes(tickangle=-45)
fig.show()

print(district_stats)
```

### Influência do Certificado Energético

```python
energy_stats = df_clean[df_clean['EnergyCertificate'].notna()].groupby('EnergyCertificate').agg({
    'Price': ['median', 'mean', 'count'],
    'PricePerM2': ['median', 'mean']
}).round(0)

fig = px.box(
    df_clean[df_clean['EnergyCertificate'].notna()],
    x='EnergyCertificate',
    y='Price',
    title='Preço por Certificado Energético',
    labels={'Price': 'Preço (€)', 'EnergyCertificate': 'Certificado'},
    color='EnergyCertificate'
)
fig.show()

print("Preço Mediano por Certificado Energético:")
print(energy_stats)

# Calcula premium A+
if ('A+',) in energy_stats.index.get_level_values(0) or 'A+' in energy_stats.index:
    a_plus_price = df_clean[df_clean['EnergyCertificate'] == 'A+']['Price'].median()
    general_price = df_clean['Price'].median()
    premium = ((a_plus_price - general_price) / general_price) * 100
    print(f"\n✨ Premium A+: {premium:.1f}% (€{a_plus_price:,.0f} vs €{general_price:,.0f})")
```

### Preço/m² por Distrito (Braga destacada)

```python
priceperm2_stats = df_clean.groupby('District')['PricePerM2'].agg(['median', 'mean', 'count']).sort_values('median', ascending=False).head(10)

fig = px.bar(
    priceperm2_stats.reset_index(),
    x='District',
    y='median',
    title='Top 10 - Preço/m² por Distrito',
    labels={'median': 'Preço/m² (€)', 'District': 'Distrito'},
    color='median',
    color_continuous_scale='RdYlGn_r'
)
fig.update_xaxes(tickangle=-45)
fig.show()

# Foco Braga
if 'Braga' in df_clean['District'].values:
    braga_price = df_clean[df_clean['District'] == 'Braga']['PricePerM2'].median()
    portugal_price = df_clean['PricePerM2'].median()
    print(f"\nBraga: €{braga_price:.0f}/m²")
    print(f"Portugal (média): €{portugal_price:.0f}/m²")
    print(f"Diferença: {((braga_price / portugal_price - 1) * 100):.1f}% mais caro")
```

### Correlações Entre Variáveis

```python
# Seleciona colunas numéricas para correlação
numeric_cols = df_clean[['Price', 'TotalArea', 'NumberOfBedrooms', 'NumberOfWC', 'PricePerM2', 'ConstructionYear']].dropna()

corr_matrix = numeric_cols.corr()

fig = px.imshow(
    corr_matrix,
    title='Matriz de Correlação',
    labels=dict(color='Correlação'),
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1
)
fig.show()

print("\nCorrelações com Preço:")
print(corr_matrix['Price'].sort_values(ascending=False))
```

## 💡 Insights Principais

1. **Distribuição de Preços**: Maioria concentrada em €84k-€395k (Q1-Q3)
2. **Tipo de Imóvel**: Casas custam ~[valor] mais que apartamentos
3. **Localização**: Braga é **30,7% mais barato** que a média nacional
